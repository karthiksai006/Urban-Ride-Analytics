```python
import pandas as pd

bangalore = pd.read_csv("cleaned_bangalore.csv")
ncr = pd.read_csv("cleaned_ncr.csv")
```

```python
bangalore["DateTime"] = pd.to_datetime(
    bangalore["Date"].astype(str) + " " + bangalore["Time"].astype(str),
    errors="coerce"
)

ncr["DateTime"] = pd.to_datetime(
    ncr["Date"].astype(str) + " " + ncr["Time"].astype(str),
    errors="coerce"
)
```

```python
combined_df = pd.concat([bangalore, ncr], ignore_index=True)
```

```python
combined_df.drop(columns=["Date", "Time"], inplace=True)
```

```python
# Hour
combined_df["Hour"] = combined_df["DateTime"].dt.hour

# Day
combined_df["Day"] = combined_df["DateTime"].dt.day

# Month
combined_df["Month"] = combined_df["DateTime"].dt.month

# Weekday
combined_df["Weekday"] = combined_df["DateTime"].dt.day_name()

# Weekend flag
combined_df["Is_Weekend"] = combined_df["Weekday"].isin(["Saturday", "Sunday"]).astype(int)
```

```python
def get_time_slot(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

combined_df["Time_Slot"] = combined_df["Hour"].apply(get_time_slot)
```
